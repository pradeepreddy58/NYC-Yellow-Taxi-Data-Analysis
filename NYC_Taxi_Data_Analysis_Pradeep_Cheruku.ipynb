### Cleaning Summary

The dataset was cleaned while preserving potentially useful trip records.

- Duplicate records: No duplicates were found.
- Invalid timestamp order: 124 records were removed because dropoff occurred before pickup.
- Negative trip duration: No negative durations remained after timestamp cleaning.
- Negative trip distance: No negative-distance records were found.
- Zero-distance trips: Retained for further analysis because they may represent unusual but potentially meaningful records.
- Zero-passenger trips: Retained for further analysis rather than automatically removing them.
- Missing values: Retained in passenger_count, RatecodeID, store_and_fwd_flag, congestion_surcharge, and Airport_fee because the affected records still contain useful trip-level information.
- Extreme trip durations: Trips longer than 24 hours were identified and inspected but were not automatically removed.
- Outliers: Extreme values will be evaluated later using IQR-based analysis.

```python
# Create trip duration in minutes

clean_df["trip_duration_minutes"] = (
    clean_df["tpep_dropoff_datetime"]
    - clean_df["tpep_pickup_datetime"]
).dt.total_seconds() / 60

print("trip_duration_minutes column created successfully.")
print(clean_df["trip_duration_minutes"].head())
```

```python
# Create pickup date

clean_df["pickup_date"] = clean_df["tpep_pickup_datetime"].dt.date

print("pickup_date column created successfully.")
print(clean_df["pickup_date"].head())
```

```python
# Create pickup hour

clean_df["pickup_hour"] = clean_df["tpep_pickup_datetime"].dt.hour

print("pickup_hour column created successfully.")
print(clean_df["pickup_hour"].head())
```

```python
# Create day of the week

clean_df["day_of_week"] = clean_df["tpep_pickup_datetime"].dt.day_name()

print("day_of_week column created successfully.")
print(clean_df["day_of_week"].head())
```

```python
# Create weekend indicator

clean_df["is_weekend"] = clean_df["day_of_week"].isin(
    ["Saturday", "Sunday"]
)

print("is_weekend column created successfully.")
print(clean_df["is_weekend"].value_counts())
```

```python
# Create average speed in miles per hour

clean_df["average_speed_mph"] = (
    clean_df["trip_distance"] /
    (clean_df["trip_duration_minutes"] / 60)
)

print("average_speed_mph column created successfully.")
print(clean_df["average_speed_mph"].head())
```

```python
# Create distance categories

clean_df["distance_category"] = pd.cut(
    clean_df["trip_distance"],
    bins=[-1, 2, 10, float("inf")],
    labels=["Short", "Medium", "Long"]
)

print("distance_category column created successfully.")
print(clean_df["distance_category"].value_counts())
```

```python
# Create duration categories

clean_df["duration_category"] = pd.cut(
    clean_df["trip_duration_minutes"],
    bins=[-1, 5, 15, 30, float("inf")],
    labels=["Very Short", "Short", "Medium", "Long"]
)

print("duration_category column created successfully.")
print(clean_df["duration_category"].value_counts())
```

```python
# Create route from pickup and drop-off location IDs

clean_df["route"] = (
    clean_df["PULocationID"].astype(str)
    + " → "
    + clean_df["DOLocationID"].astype(str)
)

print("route column created successfully.")
print(clean_df["route"].head())
```

IQR Outlier Analysis

```python
# IQR-based outlier analysis for trip distance

distance_q1 = np.percentile(distance_values, 25)
distance_q3 = np.percentile(distance_values, 75)

distance_iqr = distance_q3 - distance_q1

distance_lower_bound = distance_q1 - 1.5 * distance_iqr
distance_upper_bound = distance_q3 + 1.5 * distance_iqr

distance_outliers = (
    (clean_df["trip_distance"] < distance_lower_bound) |
    (clean_df["trip_distance"] > distance_upper_bound)
)

print("Q1:", distance_q1)
print("Q3:", distance_q3)
print("IQR:", distance_iqr)
print("Lower Bound:", distance_lower_bound)
print("Upper Bound:", distance_upper_bound)
print("Number of Distance Outliers:", distance_outliers.sum())
print(
    "Percentage of Distance Outliers:",
    (distance_outliers.mean() * 100)
)
```

**Insight:**  
Using the 1.5×IQR rule, the upper bound for trip distance is 6.28 miles. A total of 422,384 trips, representing 12.15% of the dataset, are classified as statistical distance outliers. These observations were not removed because an IQR outlier is not necessarily an invalid trip. The extreme distance values will be considered separately during anomaly investigation.

```python
# IQR-based outlier analysis for trip duration

duration_q1 = np.percentile(duration_values, 25)
duration_q3 = np.percentile(duration_values, 75)

duration_iqr = duration_q3 - duration_q1

duration_lower_bound = duration_q1 - 1.5 * duration_iqr
duration_upper_bound = duration_q3 + 1.5 * duration_iqr

duration_outliers = (
    (clean_df["trip_duration_minutes"] < duration_lower_bound) |
    (clean_df["trip_duration_minutes"] > duration_upper_bound)
)

print("Q1:", duration_q1)
print("Q3:", duration_q3)
print("IQR:", duration_iqr)
print("Lower Bound:", duration_lower_bound)
print("Upper Bound:", duration_upper_bound)
print("Number of Duration Outliers:", duration_outliers.sum())
print(
    "Percentage of Duration Outliers:",
    (duration_outliers.mean() * 100)
)
```

**Insight:**  
Using the 1.5×IQR rule, the upper bound for trip duration is 34.91 minutes. A total of 191,020 trips, representing 5.50% of the dataset, are classified as statistical duration outliers. The lower IQR bound is negative and therefore does not represent a practical duration threshold. These observations were retained for further investigation rather than automatically removed.

```python
# IQR-based outlier analysis for average speed

speed_q1 = np.percentile(speed_values, 25)
speed_q3 = np.percentile(speed_values, 75)

speed_iqr = speed_q3 - speed_q1

speed_lower_bound = speed_q1 - 1.5 * speed_iqr
speed_upper_bound = speed_q3 + 1.5 * speed_iqr

speed_outliers = (
    (clean_df["average_speed_mph"] < speed_lower_bound) |
    (clean_df["average_speed_mph"] > speed_upper_bound)
)

print("Q1:", speed_q1)
print("Q3:", speed_q3)
print("IQR:", speed_iqr)
print("Lower Bound:", speed_lower_bound)
print("Upper Bound:", speed_upper_bound)
print("Number of Speed Outliers:", speed_outliers.sum())
print(
    "Percentage of Speed Outliers:",
    (speed_outliers.mean() * 100)
)
```

**Insight:**  
Using the 1.5×IQR rule, the upper bound for average speed is 21.52 mph. A total of 270,262 trips, representing 7.78% of the dataset, are classified as statistical speed outliers. The lower IQR bound is negative and therefore does not represent a practical speed threshold. These observations were retained because statistical outliers are not necessarily invalid records and require further investigation.
